```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import accuracy_score, roc_auc_score, f1_score
```

```python
data = pd.read_csv("Customer Churn.csv")
```

```python
data
```

```python
X = data.drop('Churn', axis=1)
Y = data['Churn']
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
X_train_scaled_df
```

```python
X_test_scaled_df
```

```python
# this is the logistic regression model
```

```python
lr = LogisticRegression(max_iter=5000, random_state=42)
lr.fit(X_train_scaled_df, Y_train)
```

```python
Y_pred_lr = lr.predict(X_test_scaled_df)
Y_prob_lr = lr.predict_proba(X_test_scaled_df)[:, 1]
```

```python
print("Logistic Regression Performance:")
print("AUC:", roc_auc_score(Y_test, Y_prob_lr))
print("Accuracy:", accuracy_score(Y_test, Y_pred_lr))
print("F1 Score:", f1_score(Y_test, Y_pred_lr))
```

```python
# this is thhe decision tree classifier model
```

```python
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train_scaled_df, Y_train)
```

```python
Y_pred_dt = dt.predict(X_test_scaled_df)
Y_prob_dt = dt.predict_proba(X_test_scaled_df)[:, 1]
```

```python
print("\nDecision Tree Performance:")
print("AUC:", roc_auc_score(Y_test, Y_prob_dt))
print("Accuracy:", accuracy_score(Y_test, Y_pred_dt))
print("F1 Score:", f1_score(Y_test, Y_pred_dt))
```

```python
# this is the random forest classifier model
```

```python
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train_scaled_df, Y_train)
```

```python
Y_pred_rf = rf.predict(X_test_scaled_df)
Y_prob_rf = rf.predict_proba(X_test_scaled_df)[:, 1]
```

```python
print("\nRandom Forest Performance:")
print("AUC:", roc_auc_score(Y_test, Y_prob_rf))
print("Accuracy:", accuracy_score(Y_test, Y_pred_rf))
print("F1 Score:", f1_score(Y_test, Y_pred_rf))
```

```python
### The random forest classifier model achieved the highest performance across all metrics, with an AUC of 0.977, an Accuracy of 0.938, and an F1 Score of 0.817.
```

```python
data
```

```python
data_reg = data[data['Customer Value'] != 0]
```

```python
X = data_reg.drop(['Customer Value'], axis=1)
Y = data_reg['Customer Value']
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

```python
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
X_train_scaled_df
```

```python
X_test_scaled_df
```

```python
#the first model is a linear regression model
```

```python
lr = LinearRegression()
lr.fit(X_train_scaled_df, Y_train)
```

```python
Y_pred_lr = lr.predict(X_test_scaled_df)
```

```python
print("Linear Regression Performance:")
print("RMSE:", np.sqrt(mean_squared_error(Y_test, Y_pred_lr)))
print("MAE:", mean_absolute_error(Y_test, Y_pred_lr))
print("MAPE:", mean_absolute_percentage_error(Y_test, Y_pred_lr))
```

```python
#the second model is a decision tree regressor
```

```python
dt = DecisionTreeRegressor(random_state=42)
dt.fit(X_train_scaled_df, Y_train)
```

```python
Y_pred_dt = dt.predict(X_test_scaled_df)
```

```python
print("\nDecision Tree Regressor Performance:")
print("RMSE:", np.sqrt(mean_squared_error(Y_test, Y_pred_dt)))
print("MAE:", mean_absolute_error(Y_test, Y_pred_dt))
print("MAPE:", mean_absolute_percentage_error(Y_test, Y_pred_dt))
```

```python
#the third model is a random forest regressor
```

```python
rf = RandomForestRegressor(random_state=42)
rf.fit(X_train_scaled_df, Y_train)
```

```python
Y_pred_rf = rf.predict(X_test_scaled_df)
```

```python
print("\nRandom Forest Regressor Performance:")
print("RMSE:", np.sqrt(mean_squared_error(Y_test, Y_pred_rf)))
print("MAE:", mean_absolute_error(Y_test, Y_pred_rf))
print("MAPE:", mean_absolute_percentage_error(Y_test, Y_pred_rf))
```

```python
### The random forest regressor performed the best, getting the lowest error rates across all metrics, with a 31.779 RMSE, a 11.029 MAE, and a 0.050 MAPE.
```
